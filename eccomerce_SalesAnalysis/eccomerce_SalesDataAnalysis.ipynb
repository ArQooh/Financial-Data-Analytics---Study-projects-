```python
df['Review Rating'] = df['Review Rating'].fillna(value=0, inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.replace(" ", "_")
df.columns = df.columns.str.lower()
```

```python
df = df.rename(columns={'purchase_amount_(usd)' : 'sales'})
```

```python
df.columns
```

```python
myLabels = ['Young(18-29)', 'Adult(30-59)', 'Senior(60+)']
df['customer_age_group'] = pd.qcut(df['age'], q=[0, 0.25, 0.8, 1], labels=myLabels)
```

```python
df[df['age'] > 30].head(10)
```

```python
df['frequency_of_purchases'] = df['frequency_of_purchases'].replace({
    'Fortnightly': 'Fortnightly(14 days)',
    'Bi-Weekly': 'Bi-Weekly(14 days)',
    'Weekly': 'Weekly(7 days)',
    'Quarterly': 'Quarterly(90 days)',
    'Annually': 'Annually(365 days)'
}, inplace=True)
```

```python
df = df.drop('promo_code_used', axis= 1)
```

```python
from sqlalchemy import create_engine

# MySQL connection
username = "root"
password = "1234"
host = "localhost"
port = "3306"
database = "db_sales_analysis"

engine = create_engine(f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}")

# Write DataFrame to MySQL
table_name = "customer_behaviour"   # choose any table name
df.to_sql(table_name, engine, if_exists="replace", index=False)

# Read back sample
pd.read_sql("SELECT * FROM customer_behaviour LIMIT 5;", engine)
```
